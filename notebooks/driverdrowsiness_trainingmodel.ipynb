```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import ModelCheckpoint
import os
```

```python
# Paths folders in Drive
train_dir = '/content/drive/MyDrive/Drivers Drowsiness Project/dataset_new/train'
test_dir = '/content/drive/MyDrive/Drivers Drowsiness Project/dataset_new/test'
model_save_path = '/content/drive/MyDrive/Drivers Drowsiness Project/models/'

# Image & training config
img_size = (64, 64)
batch_size = 32
num_classes = 4  # ['Closed', 'Open', 'Yawn', 'No Yawn']
epochs = 10
```

```python
datagen = ImageDataGenerator(
    rescale=1.0/255,
    validation_split=0.1,
    horizontal_flip=True
)

train_data = datagen.flow_from_directory(
    train_dir,
    target_size=img_size,
    batch_size=batch_size,
    color_mode='grayscale',
    class_mode='categorical',
    subset='training'
)

val_data = datagen.flow_from_directory(
    train_dir,
    target_size=img_size,
    batch_size=batch_size,
    color_mode='grayscale',
    class_mode='categorical',
    subset='validation'
)
```

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(64, 64, 1)),
    MaxPooling2D(2, 2),
    Dropout(0.1),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation='relu', padding='same'),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(num_classes, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
checkpoint = ModelCheckpoint(
    filepath='/content/drive/MyDrive/Drivers Drowsiness Project/models/eye_yawn_model_v1.h5',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max',
    verbose=1
)
```

```python
steps_per_epoch = train_data.n
val_steps = val_data.n

history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=epochs,
    steps_per_epoch=steps_per_epoch,
    validation_steps=val_steps,
    callbacks=[checkpoint]
)
```

```python
print("Training complete. Model saved to:", model_save_path)
```
